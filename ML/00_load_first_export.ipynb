Нужен `pandas`. Параметры карточек уже в отдельных колонках. Архив и паспорт для загрузки не нужны.


```python
import pandas as pd
from pathlib import Path

# Работает при запуске из корня репозитория и из папки ML.
file = Path("Data/first_export_2026-10-21.csv")
if not file.exists():
    file = Path("..") / file

df = pd.read_csv(file, dtype={"id": "string"})
df.head()
```
